```python


import pandas as pd

# Загрузка данных
df = pd.read_csv('denoising_results_for_ten_pics.csv')

# Функция для нахождения лучших результатов по PSNR и SSIM для каждой картинки и типа шума
def find_best_results(df):
    results = []
    images = df['image'].unique()
    noise_types = df['noise_type'].unique()

    for image in images:
        for noise_type in noise_types:
            subset = df[(df['image'] == image) & (df['noise_type'] == noise_type)]

            # Лучший PSNR
            best_psnr = subset.loc[subset['psnr'].idxmax()]
            results.append({
                'image': image,
                'noise_type': noise_type,
                'metric': 'PSNR',
                'denoising_method': best_psnr['denoising_method'],
                'strength': best_psnr['strength'],
                'mu': best_psnr['mu'],
                'value': best_psnr['psnr'],
                'ssim': best_psnr['ssim'],
                'output_path': best_psnr['output_path']
            })

            # Лучший SSIM
            best_ssim = subset.loc[subset['ssim'].idxmax()]
            results.append({
                'image': image,
                'noise_type': noise_type,
                'metric': 'SSIM',
                'denoising_method': best_ssim['denoising_method'],
                'strength': best_ssim['strength'],
                'mu': best_ssim['mu'],
                'value': best_ssim['ssim'],
                'psnr': best_ssim['psnr'],
                'output_path': best_ssim['output_path']
            })
    results = pd.DataFrame(results)
    return results

# Получаем лучшие результаты
best_results = find_best_results(df)
pd.set_option('display.max_rows', None)
# Выводим результаты
best_results[['image', 'noise_type', 'metric', 'denoising_method', 'strength', 'mu', 'value', 'ssim', 'psnr']]
```
